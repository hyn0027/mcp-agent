```python
path = "generated_tasks_shuffled.json"

with open(path, "r") as f:
    data = f.read()
```

```python
# json to csv
import json
import csv

data = json.loads(data)

columns = ["id", "task", "policy", "task_goal", "additional_details", "explanation"]

with open("generated_tasks_shuffled.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(columns)
    for item in data:
        row = [item.get(col, "") for col in columns]
        writer.writerow(row)
```
